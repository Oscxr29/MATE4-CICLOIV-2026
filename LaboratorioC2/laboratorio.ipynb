```python
import math
import pandas as pd


f = lambda x: x**2 - 100*x*math.sin(x) + (10*math.sin(x))**2
df = lambda x: 2*x - 100*(math.sin(x) + x*math.cos(x)) + 200*math.sin(x)*math.cos(x)

def Biseccion(func, a, b, tol, maxiter=50):
    iter = 0
    error = tol + 1
    lista_x = []
    lista_e = []

    while error > tol and iter < maxiter:
        x = (a + b) / 2
        error = abs(b - a) / 2
        iter += 1

        if func(a) * func(x) < 0:
            b = x
        else:
            a = x

        lista_x.append(x)
        lista_e.append(error)

    tabla = pd.DataFrame({'Iteraciones': lista_x, 'Error': lista_e})
    return x, iter, error, tabla
```

```python
a, b = 2.1, 3.2
tol = 1e-3
maxiter = 30

x, iters, error, tabla_f = Biseccion(f, a, b, tol, maxiter)

print("La solución obtenida fue %s" % x)
print("Las iteraciones realizadas fueron %s" % iters)
print("El error obtenido fue %s" % error)
display(tabla_f)
```

```python
a, b = -3, -2
tol = 1e-3
maxiter = 30

x, iters, error, tabla_df = Biseccion(df, a, b, tol, maxiter)

print("La solución obtenida fue %s" % x)
print("Las iteraciones realizadas fueron %s" % iters)
print("El error obtenido fue %s" % error)
display(tabla_df)
```

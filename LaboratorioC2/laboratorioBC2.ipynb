```python
import pandas as pd

p = lambda x: 10*x**4 - 31*x**3 + 18*x**2 + 2*x - 1
dp = lambda x: 40*x**3 - 93*x**2 + 36*x + 2

def Newton(p, dp, x0, tol, maxiter):
    x = x0
    for iters in range(maxiter):
        fx = p(x)
        dfx = dp(x)
        if dfx == 0: break
        x_new = x - fx / dfx
        error = abs(x_new - x)
        if error < tol:
            return x_new, iters + 1, error
        x = x_new
    return x, maxiter, error

def Newtontabla(p, dp, x0, tol, maxiter):
    x = x0
    hx, he = [], []
    for i in range(maxiter):
        fx = p(x)
        dfx = dp(x)
        if dfx == 0: break
        x_new = x - fx / dfx
        error = abs(x_new - x)
        hx.append(x_new)
        he.append(error)
        if error < tol:
            break
        x = x_new
    return pd.DataFrame({'Iteraciones': hx, 'Error': he})
```

```python
x0 = -0.3
tol = 1e-7
maxiter = 30

r1, iters, error = Newton(p, dp, x0, tol, maxiter)

print("La solución obtenida para r1 es x = %s" % r1)
print("El numero de iteraciones fue de %s" % iters)
print("El error obtenido fue %s" % error)
```

```python
Newtontabla(p, dp, x0, tol, maxiter)
```

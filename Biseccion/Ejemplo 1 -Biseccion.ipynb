```python
import math
import numpy as np
import pandas as pd
```

```python
def Biseccion(f,a,b,tol,maxiter):
    iters = 0
    error = tol+1

    while iters<maxiter and  error>tol:
        x = (a+b)/2
        error = (b-a)/2

        if f(a)*f(x)>0:
            a = x
        else:
            b = x
        iters = iters+1
    return [x,iters,error]
```

```python
# Ejemplo 1

f = lambda x: x**3+4*x**2-10
a = 1
b = 2
tol = 1e-3 #(10^(-3))
maxiter = 30
```

```python
[x,iters,error] = Biseccion(f,a,b,tol,maxiter)
```

```python
print("La solucion obtenida fue %s" %x)
print("La iteraciones realizadas fueron %s" %iters)
print("El error obtenido fue %s" %error)
```

## Ejemplo

Una partícula parte del reposo sobre un plano inclinado uniforme. Al final de $t$ segundos, la posición del objeto está dada por
$$x(t)=-\dfrac{g}{2\omega^2} \left(\frac{e^{\omega t}-e^{-\omega t}}{2} -\sin (\omega t)\right)$$
Suponga que la partícula se desplazó $1.7$ pies en dirección negativa en $1 \ s$. Encontrar con una exactitud de $10^{-5}$, la rapidez $\omega$. Suponga que $g=32.17 \ pies/s^2$ y que $\omega \in [0,1]$

```python
import math
import pandas as pd
```

```python
def fun(w):
    g = 32.17
    t = 1
    y = -g/(2*w**2)*((math.exp(w*t)-math.exp(-w*t))/2-math.sin(w*t))-1.7
    return y
```

```python
a = -1
b = 0
tol = 1e-5
maxiter = 30
```

```python
def Biseccion(f, a, b, tol, maxiter):
    iters = 0
    error = tol + 1

    while iters < maxiter and error > tol:
        x = (a + b) / 2
        error = abs(b - a) / 2

        if f(a) * f(x) > 0:
            a = x
        else:
            b = x

        iters += 1

    return x, iters, error
```

```python
def Bisecciontabla(f, a, b, tol, maxiter):
    iters = 0
    error = tol + 1
    filas = []

    while iters < maxiter and error > tol:
        x = (a + b) / 2
        error = abs(b - a) / 2
        filas.append([iters + 1, a, b, x, error, f(x)])

        if f(a) * f(x) > 0:
            a = x
        else:
            b = x

        iters += 1

    tabla = pd.DataFrame(filas, columns=["Iteración", "a", "b", "x", "Error", "f(x)"])
    return tabla
```

```python
## Obtenemos los reusltados

x, iters, error = Biseccion(fun, a, b, tol, maxiter)

```

```python
print(f"La solución obtenida es  = {x}")
print(f"Iteraciones: {iters}")
print(f"Error: {error}")
```

```python
tabla = Bisecciontabla(fun, a, b, tol, maxiter)
tabla.head(20)  # Muestra las primeras 20 filas de la tabla
```

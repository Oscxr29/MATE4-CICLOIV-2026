**Ejercicio 4. (*) (2 pts).** La deflexión vertical de una viga está descrita mediante la siguiente función:

$$y(x)=\frac{w_{0}}{120EIL}(-x^{5}+2L^{2}x^{3}-L^{4}x)$$

Donde se tiene que $L=2\text{ m}$, $w_{0}=12\times10^{7}\text{ N/m}$, $E=2\times10^{11}\text{ N/m}^{2}$ e $I=5\times10^{-4}\text{ m}^{4}$.
Utilizar el Método de Newton para determinar la posición a lo largo de la viga donde la deflexión vertical es de $-0.003\text{ m}$ y una tolerancia de $10^{-6}$.

```python
import numpy as np

# variables globales de la viga
L = 2.0
w0 = 12e7
E = 2e11
I = 5e-4
K = w0 / (120 * E * I * L)

# f(x) y su derivada
def f(x):
    return K * (-x**5 + 2*(L**2)*(x**3) - (L**4)*x) + 0.003

def df(x):
    return K * (-5*x**4 + 6*(L**2)*(x**2) - (L**4))

def newton_viga(x0, tol):
    x = x0
    iteracion = 0

    while True:
        x_new = x - f(x)/df(x)
        error = abs(x_new - x)

        print(f"Iteración {iteracion}: x = {x_new:.6f}, error = {error:.2e}")

        if error < tol:
            return x_new

        x = x_new
        iteracion += 1

raiz = newton_viga(1.0, 1e-6)
print(f"\nLa posición en la viga es x = {raiz:.6f} m")
```

**Ejercicio 5. (*)** Se define el método MXY mediante la siguiente expresión:

$$y_{k}=x_{k}-\frac{2}{3}\frac{f(x_{k})}{f^{\prime}(x_{k})}$$
$$x_{k+1}=x_{k}-\frac{f(x_{k})}{f^{\prime}(y_{k})}\cdot\frac{5f^{\prime}(y_{k})-f^{\prime}(x_{k})}{4f^{\prime}(y_{k})}$$

a) (1.0 pts). Programar el método del MXY en Python.
b) (1.0 pts). Utilizar el método anterior para resolver la ecuación $e^{x}\cos(x)-x^{3}\sin(x)-1=0$ usando una estimación de $x_{0}=3$ y una tolerancia de $10^{-7}$. Comparar los resultados con los métodos de Newton, Traub y Ostrowski.

```python
import numpy as np

def f(x):
    return np.exp(x)*np.cos(x) - (x**3)*np.sin(x) - 1

def df(x):
    return np.exp(x)*(np.cos(x) - np.sin(x)) - (x**2)*(3*np.sin(x) + x*np.cos(x))

def metodo_newton(x0, tol):
    x = x0
    iters = 0
    while True:
        x_new = x - f(x)/df(x)
        if abs(x_new - x) < tol: return x_new, iters + 1
        x = x_new
        iters += 1

def metodo_traub(x0, tol):
    x = x0
    iters = 0
    while True:
        fx = f(x)
        dfx = df(x)
        y = x - fx/dfx
        x_new = y - f(y)/dfx
        if abs(x_new - x) < tol: return x_new, iters + 1
        x = x_new
        iters += 1

def metodo_ostrowski(x0, tol):
    x = x0
    iters = 0
    while True:
        fx = f(x)
        dfx = df(x)
        y = x - fx/dfx
        fy = f(y)
        x_new = y - (fx / (fx - 2*fy)) * (fy / dfx)
        if abs(x_new - x) < tol: return x_new, iters + 1
        x = x_new
        iters += 1

def metodo_mxy(x0, tol):
    x = x0
    iters = 0
    while True:
        fx = f(x)
        dfx = df(x)
        y = x - (2/3) * (fx / dfx)
        dfy = df(y)
        x_new = x - (fx / dfy) * ((5*dfy - dfx) / (4*dfy))
        if abs(x_new - x) < tol: return x_new, iters + 1
        x = x_new
        iters += 1
```

```python
x0 = 3.0
tolerancia = 1e-7

raiz_newton, iter_newton = metodo_newton(x0, tolerancia)
raiz_traub, iter_traub = metodo_traub(x0, tolerancia)
raiz_ostro, iter_ostro = metodo_ostrowski(x0, tolerancia)
raiz_mxy, iter_mxy = metodo_mxy(x0, tolerancia)
```

```python
print("COMPARACIÓN DE LOS MÉTODOS")
print("-" * 55)
print(f"{'Método':<15} | {'Raíz Encontrada':<15} | {'Iteraciones'}")
print("-" * 55)
print(f"{'Newton':<15} | {raiz_newton:<15.8f} | {iter_newton}")
print(f"{'Traub':<15} | {raiz_traub:<15.8f} | {iter_traub}")
print(f"{'Ostrowski':<15} | {raiz_ostro:<15.8f} | {iter_ostro}")
print(f"{'MXY':<15} | {raiz_mxy:<15.8f} | {iter_mxy}")
print("-" * 55)
```

```python
import math
import pandas as pd
```

```python
def Newton(f, df, x0, tol, maxiter):
    iters = 0
    error = tol + 1

    while error > tol and iters < maxiter:
        x = x0 - f(x0) / df(x0)
        error = abs(x - x0)
        x0 = x
        iters += 1

    return x, iters, error
```

## Ejercicio 

Resolver $x^3-x-1=0$ usando el metodo de newton con una estimacion einicial de $x_0=-2$ y una tolerancia de $10^{-6}$

```python
#definimos f

f = lambda x: x**3-x-1

#Obtenemos la derivada de f (f'(x) = 3x^2-1)

df= lambda x: 3*x**2-1
x0= 1.5
tol= 1e-6
maxiter= 30
```

```python
x,iters,error = Newton(f, df, x0, tol, maxiter)

print(f"La solución obtenida es  = {x}")
print(f"Iteraciones: {iters}")
print(f"Error: {error}")
```

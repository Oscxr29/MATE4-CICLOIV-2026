```python
import pandas as pd

def NewtonTabla(f, df, x0, tol, maxiter):
    iters = 0
    error = tol + 1
    lista_e = []
    lista_x = []

    while error > tol and iters < maxiter:
        x = x0 - f(x0) / df(x0)
        error = abs(x - x0)
        x0 = x
        iters += 1

        lista_e.append(error)
        lista_x.append(x)

    tabla = pd.DataFrame({
        "Iteraciones": lista_x,
        "Errores": lista_e
    })

    pd.set_option("display.precision", 10)

    return tabla, x, iters, error
```

## Ejemplo 

Usar el metodo de newton para resolver la ecuacion $x^3-3x-5=0$ con una tolerancia de $10^{-6}$ con una estimacion inicial de $x_0=2$

```python
#Introducir f

f= lambda x: x**3-3*-5

# Introducir la dervivda de f
df= lambda x: 3*x**2-3
x0= 2
tol = 1e-6
maxiter = 30
```

```python
NewtonTabla(f,df,x0,tol,maxiter)
```
